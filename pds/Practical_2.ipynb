# Practical 2: Convert the Unstructured Log Data into a Structured Dataset

**Objective:** Parse log data into structured tabular form.

- Use regex to extract: IP address, date/time, request type, status code, resource, bytes sent, referrer, user-agent
- Store data in a Pandas DataFrame

```python
import re
import pandas as pd

with open("logs/access.log", "r") as file:
    log_lines = file.readlines()

print("Total log entries:", len(log_lines))
```

## Regex cheat sheet

| Regex | Meaning |
|---|---|
| `.` | Any character |
| `\d` | Any digit |
| `\w` | Word character |
| `\s` / `\S` | Whitespace / non-whitespace |
| `+` / `*` / `?` | One-or-more / zero-or-more / zero-or-one |
| `[]` / `[^]` | Character set / negated set |
| `()` | Capturing group |
| `^` / `$` | Start / end of string |

```python
# Regex pattern for Apache/Nginx Combined Log Format
pattern = r'(\S+) \S+ \S+ \[([^\]]+)\] "(\S+) (\S+) ([^"]+)" (\d+) (\d+) "([^"]*)" "([^"]*)"'
```

## Parse each line into structured fields

```python
data = []

for line in log_lines:
    match = re.search(pattern, line)
    if match:
        data.append({
            "IP Address": match.group(1),
            "Date/Time": match.group(2),
            "Request Type": match.group(3),
            "Resource": match.group(4),
            "Protocol": match.group(5),
            "Status Code": int(match.group(6)),
            "Bytes Sent": int(match.group(7)),
            "Referrer": match.group(8),
            "User Agent": match.group(9)
        })

print(f"Parsed {len(data)} of {len(log_lines)} lines successfully")
```

## Build the DataFrame

```python
df = pd.DataFrame(data)
df.head(10)
```
