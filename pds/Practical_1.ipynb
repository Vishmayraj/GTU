# Practical 1: Load and Explore the Unstructured Access Log Data

**Objective:** Learn how to import and explore raw server access logs (Apache/Nginx format).

- Load a sample `access.log` file using Python
- Understand the typical log structure (IP, timestamp, request, status, size, user-agent, etc.)
- Extract 5-10 random lines to manually inspect the pattern
- Identify useful fields for analysis

```python
import random

with open('logs/access.log', 'r') as file:
    log_lines = file.readlines()

print("Total number of log lines:", len(log_lines))
```

## First log entry

```python
print(log_lines[0])
```

## Useful fields in Access Logs

A standard Apache/Nginx **Combined Log Format** entry contains:

| Field | Example | Useful for |
|---|---|---|
| IP Address | `192.168.1.10` | tracking source, rate-limiting, geo |
| Timestamp | `[12/Aug/2026:10:15:32 +0530]` | time-series analysis, brute-force detection |
| HTTP Method | `GET` / `POST` | request type analysis |
| Resource | `/index.html` | attack pattern detection (SQLi, traversal) |
| Protocol | `HTTP/1.1` | sanity check |
| Status Code | `200`, `404`, `500` | anomaly detection |
| Bytes Sent | `4521` | payload size anomalies |
| Referrer | `https://google.com` | traffic source |
| User-Agent | `Mozilla/5.0 ...` | bot vs. real browser detection |

## Random sample of 5 log entries

```python
random_logs = random.sample(log_lines, 5)

print("Random Log Entries:")
for entry in random_logs:
    print(entry.strip())
```

## Observations

- Each line follows the Combined Log Format: `IP - - [timestamp] "METHOD path PROTOCOL" status bytes "referrer" "user-agent"`
- Fields are space-separated, but the request, timestamp, referrer, and user-agent are each wrapped in `[]` or `""` since they may contain spaces themselves — this is why regex (not `.split()`) will be needed to parse them reliably in Practical 2.
- The most analytically useful fields for anomaly/attack detection later are: **Resource** (for SQLi/traversal patterns), **Status Code** (for error/anomaly spikes), and **IP + Timestamp** (for brute-force / rate-based detection).
